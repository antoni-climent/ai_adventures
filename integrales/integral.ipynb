```python
data = pd.read_csv("data.csv")
```

```python
# Make histogram
def make_histogram(data:np.array, num_div):
    min = data.min()
    max = data.max()
    step_size = (max - min)/num_div
    print(min, max)
    histo = np.array([0]*num_div)
    for el in data:
        index = int((el - min)/step_size)
        histo[index-1]+=1
    bin_edges = np.arange(min, max, step_size)
    return histo, bin_edges
    
def plot_histogram(counts, bin_edges):
    plt.plot(bin_edges, counts)
    plt.savefig("plot.png")


counts, bin_edges = make_histogram(np.array(data["texture_mean"]), 100)
# Plot histogram
plot_histogram(counts, bin_edges)
plot_histogram(np.log(counts), bin_edges)
```

```python
# Fit the data with gaussian function:
mu = 0.00001
de = 5

x_axis = np.arange(-10,10,0.01)

def gaus(x):
    return (1/(de*np.sqrt(2*np.pi)))*np.e**(-1/2*((x-mu)/de)**2)

def der_gaus(x):
    return -(x-mu)/de**2 * gaus(x)

def loss(count, bin_edge):
    pred = gaus(bin_edge)
    l = (pred - count)**2 
    return l


```
